**MODEL IMPLEMENTATION**

```python

class KNN:
  def __init__(self , k =5 ):
    self.k = k # Number of nearest neighbors to consider
  def euclidean_distance(self,x1,x2): # Calculate Euclidean distance between two data points
    return np.linalg.norm(x1-x2)
  def fit(self,x_train,y_train):# Store training data and corresponding labels
    self.x_train = x_train
    self.y_train = y_train

  def predict(self,X): # List to store predictions for each test point
    predictions = []
    for x in X:
      prediction = self._predict(x) # Predict label for one data point
      predictions.append(prediction)
    return np.array(predictions)# Convert list to NumPy array

  def _predict(self,X):
    distances = [] # Store distances from X to all training points

    # Compute distance between test point and each training point

    for x_train in self.x_train :
      distance = self.euclidean_distance(X,x_train)
      distances.append(distance)
    distances = np.array(distances)

    # Get indices of k smallest distances

    k_idxs = np.argsort(distances)[:self.k]


    # Get labels of the k nearest neighbors

    labels = self.y_train[k_idxs]
    labels = list(labels)

    # Find the most frequent label (majority voting)

    most_occ_value = max(labels , key = list(labels).count)
    return most_occ_value
```

```python
def train_test_split(X,y,random_state = 42, test_size = 0.2):
  # Total number of samples in the dataset
  n_samplen = X.shape[0]
  np.random.seed(random_state)   # Set random seed to get the same split every time (reproducibility)
  shuffled_indices = np.random.permutation(n_samplen)   # Generate a random permutation of indices
  test_set_size = int(n_samplen * test_size)   # Calculate number of samples in the test set
    # Select indices for test and train sets
  test_indices = shuffled_indices[:test_set_size]
  train_indices = shuffled_indices[test_set_size:]
  # Split the data using the indices
  x_train , x_test = X[train_indices] , X[test_indices]
  y_train , y_test = y[train_indices] , y[test_indices]

  return x_train , x_test , y_train , y_test   # Return training and testing data
```

```python
X_train, X_test, y_train, y_test = train_test_split(X.values, Y.values, test_size = 0.2, random_state=42) #split the  data into traing and validating
```

```python
model = KNN(7) # Create a KNN model with k = 7 nearest neighbors
model.fit(X_train , y_train) # Store the training data and labels in the model
```

```python
def compute_accuracy(y_true, y_pred):
  y_true = y_true.flatten()   # Convert true labels to a 1D array (if not already)
  total_samples = len(y_true)   # Total number of samples
  correct_predictions = np.sum(y_true == y_pred)  # Count how many predictions match the true labels
  return (correct_predictions / total_samples)  # Calculate and return accuracy
```

```python
predictions = model.predict(X_test) # Predict class labels for the test dataset
accuracy = compute_accuracy(y_test, predictions) # Calculate accuracy by comparing true labels with predicted labels
print(f"Accuracy: {accuracy}") # Print the model accuracy
```

**SKLEARN IMPLEMENTATION**

```python
from sklearn.neighbors import KNeighborsClassifier # Import KNN classifier from scikit-learn
skmodel = KNeighborsClassifier(n_neighbors=7) # Create a KNN model with 7 nearest neighbors
skmodel.fit(X_train, y_train) # Train (fit) the model using training data and labels
```
