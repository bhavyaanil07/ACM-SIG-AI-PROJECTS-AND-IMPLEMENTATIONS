```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("StudentsPerformance.csv")
data.head()
```

```python
data.info()
data.describe()
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.describe(include='object')
```

```python
data_encoded = pd.get_dummies(data, drop_first=True)
```

```python
data['avg_score'] = data[['math score', 'reading score', 'writing score']].mean(axis=1)
```

```python
sns.histplot(data['math score'], kde=True)
plt.title("Distribution of Math Scores")
plt.show()
```

```python
sns.boxplot(x='test preparation course', y='avg_score', data=data)
plt.title("Test Preparation Course vs Average Score")
plt.show()
```

```python
sns.boxplot(x='gender', y='math score', data=data)
plt.title("Gender vs Math Score")
plt.show()
```

```python
sns.boxplot(x='parental level of education', y='avg_score', data=data)
plt.xticks(rotation=45)
plt.title("Parental Education vs Average Score")
plt.show()
```

```python
sns.heatmap(
    data[['math score','reading score','writing score','avg_score']].corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation Between Scores")
plt.show()
```
